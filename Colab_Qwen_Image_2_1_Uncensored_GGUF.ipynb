```python
# Refer https://huggingface.co/abenzerps/Qwen-Image-2.1-Uncensored-GGUF

import os

# Downloads the ingredients
ingredients = [
    {
        "url": "https://huggingface.co/abenzerps/Qwen-Image-2.1-Uncensored-GGUF/resolve/main/qwen-image-2.1-UC-Q4_K_M.gguf",
        "path": f"{BASE}/models/diffusion_models",
        "name": "qwen-image-2.1-UC-Q4_K_M.gguf",
    },
    {
        "url": "https://huggingface.co/abenzerps/Qwen-Image-2.1-Uncensored-GGUF/resolve/main/text_encoders/qwen3vl_8b_int8_convrot.safetensors",
        "path": f"{BASE}/models/text_encoders",
        "name": "qwen3vl_8b_int8_convrot.safetensors"
    },
    {
        "url": "https://huggingface.co/abenzerps/Qwen-Image-2.1-Uncensored-GGUF/resolve/main/vae/qwen_image_2.1_vae_bf16.safetensors",
        "path": f"{BASE}/models/vae",
        "name": "qwen_image_2.1_vae_bf16.safetensors",
    }
]

force_refresh = False  # 🔁 Set to True to redownload even if file exists

for item in ingredients:

  # ✅ Construct full path
  model_path = os.path.join(item['path'], item['name'])

  # ✅ Ensure model folder exists
  os.makedirs(item['path'], exist_ok=True)
  model_url = item['url']

  # ✅ Logic to download or refresh
  if not os.path.exists(model_path):
      print(f"⬇️ item {model_path} not found. Downloading...")
      !wget -O "{model_path}" "{model_url}"
      print("✅ Model downloaded")
  elif force_refresh:
      print(f"🔁 Force refresh is enabled. Re-downloading item {model_path}...")
      !wget -O "{model_path}" "{model_url}"
      print("✅ Model refreshed")
  else:
      print(f"✅ {model_path} already exists. Skipping download.")
```

```python
import subprocess
import time
import re
import requests

print("🚀 Launching Cloudflare Quick Tunnel...")

# 1. Start cloudflared and redirect logs to a file
with open("cloudflared.log", "w") as log_file:
    cf_process = subprocess.Popen(
        ['cloudflared', 'tunnel', '--url', 'http://localhost:8188'],
        stdout=log_file,
        stderr=log_file
    )

public_url = None

# 2. Extract the dynamically generated URL from the logs
for i in range(15):
    try:
        with open("cloudflared.log", "r") as f:
            log_content = f.read()
            # Match the trycloudflare URL generated in the banner
            match = re.search(r'https://[a-zA-Z0-9-]+\.trycloudflare\.com', log_content)
            if match:
                public_url = match.group(0)
                break
    except Exception:
        pass

    print(f"⏳ Attempt {i+1}/15: Waiting for Cloudflare to assign a URL...")
    time.sleep(2)

# 3. Verify that the tunnel is actually live and routing traffic
if public_url:
    print(f"🔗 Found URL: {public_url}")
else:
    print("❌ Failed to retrieve the trycloudflare URL from the logs. Check !cat cloudflared.log")
```
